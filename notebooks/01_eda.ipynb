```python
import sys

import matplotlib.pyplot as plt
import pandas as pd

from src import config
from src.data_loader import inspect_columns, load_and_split

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

print("Python:", sys.version.split()[0])
print("Interpreter:", sys.executable)
print("Dataset:", config.DATA_PATH)
```

```python
columns = inspect_columns()

print("Kolom CSV:")
for position, column in enumerate(columns, start=1):
    print(f"{position:02d}. {column}")
```

```python
split = load_and_split()

X_train = split.X_train.copy()
y_train = split.y_train.copy()

df_train = X_train.copy()
df_train[config.TARGET_NAME] = y_train

print("Jumlah baris train:", len(X_train))
print("Jumlah fitur:", X_train.shape[1])
print("Target: 0=non-default, 1=default")
```

```python
display(df_train.head(10))

df_train.info()

display(
    pd.DataFrame({
        "dtype": X_train.dtypes.astype(str),
        "jenis_eda": [
            "categorical_candidate"
            if column in config.EDA_CATEGORICAL_COLUMNS
            else "numeric"
            for column in X_train.columns
        ],
    })
)
```

```python
counts = y_train.value_counts().reindex([0, 1], fill_value=0)

class_summary = pd.DataFrame({
    "label": [0, 1],
    "kelas": ["Non-default", "Default"],
    "jumlah": counts.to_numpy(),
    "persentase": (
        counts / len(y_train) * 100
    ).round(2).to_numpy(),
})

display(class_summary)

ax = class_summary.plot.bar(
    x="kelas",
    y="jumlah",
    color=["#397E41", "#BA1A1A"],
    legend=False,
    rot=0,
    figsize=(6, 4),
)

ax.set_title("Distribusi kelas pada train")
ax.set_xlabel("Kelas")
ax.set_ylabel("Jumlah nasabah")

plt.tight_layout()
plt.show()
```

```python
blank_counts = X_train.apply(
    lambda column: column.map(
        lambda value: isinstance(value, str) and not value.strip()
    ).sum()
)

feature_summary = pd.DataFrame({
    "dtype": X_train.dtypes.astype(str),
    "unique_non_missing": X_train.nunique(dropna=True),
    "missing_count": X_train.isna().sum(),
    "missing_pct": (X_train.isna().mean() * 100).round(2),
    "blank_string_count": blank_counts,
})

display(
    feature_summary.sort_values(
        "missing_count",
        ascending=False,
    )
)
```

```python
unique_by_class = (
    df_train
    .groupby(config.TARGET_NAME)[list(config.FEATURE_COLUMNS)]
    .nunique(dropna=True)
    .T
    .rename(columns={
        0: "unique_non_default",
        1: "unique_default",
    })
)

display(unique_by_class)
```

```python
duplicate_summary = pd.DataFrame({
    "pemeriksaan": [
        "Baris tambahan identik: fitur dan target",
        "Baris tambahan dengan fitur identik",
        "Semua baris dalam kelompok fitur duplikat",
    ],
    "jumlah": [
        int(df_train.duplicated().sum()),
        int(X_train.duplicated().sum()),
        int(X_train.duplicated(keep=False).sum()),
    ],
})

display(duplicate_summary)

duplicate_mask = X_train.duplicated(keep=False)

if duplicate_mask.any():
    duplicate_rows = df_train.loc[duplicate_mask]

    display(duplicate_rows.head(20))

    label_counts = (
        duplicate_rows
        .groupby(
            list(config.FEATURE_COLUMNS),
            dropna=False,
            observed=True,
        )[config.TARGET_NAME]
        .nunique()
    )

    print(
        "Kelompok fitur identik dengan dua label berbeda:",
        int((label_counts > 1).sum()),
    )
else:
    print("Tidak ditemukan kombinasi fitur duplikat pada train.")
```

```python
for column in config.EDA_CATEGORICAL_COLUMNS:
    summary = (
        df_train
        .groupby(column, dropna=False, observed=True)[config.TARGET_NAME]
        .agg(
            jumlah="size",
            jumlah_default="sum",
            default_rate="mean",
        )
        .reset_index()
    )

    summary["jumlah_non_default"] = (
        summary["jumlah"] - summary["jumlah_default"]
    )

    summary["default_rate_pct"] = (
        summary.pop("default_rate") * 100
    ).round(2)

    mapping = config.CATEGORY_LABELS.get(column)

    if mapping is not None:
        summary["arti_kategori"] = summary[column].map(mapping)
        summary["arti_kategori"] = summary["arti_kategori"].fillna(
            "Kode belum dijelaskan dalam mapping"
        )

    print(f"\nFitur: {column}")
    display(summary.sort_values(column))
```

```python
numeric_columns = list(config.NUMERIC_COLUMNS)

display(
    X_train[numeric_columns]
    .describe(percentiles=[0.01, 0.25, 0.50, 0.75, 0.99])
    .T
    .round(2)
)

numeric_by_class = (
    df_train
    .groupby(config.TARGET_NAME)[numeric_columns]
    .agg(["mean", "median", "std", "min", "max"])
    .rename(index={
        0: "Non-default",
        1: "Default",
    })
)

display(numeric_by_class.round(2))
```

```python
for column, mapping in config.CATEGORY_LABELS.items():
    undocumented = (
        X_train.loc[
            X_train[column].notna()
            & ~X_train[column].isin(mapping),
            column,
        ]
        .value_counts()
        .rename_axis("kode")
        .reset_index(name="jumlah")
    )

    print(f"\nKode di luar mapping: {column}")

    if undocumented.empty:
        print("Tidak ditemukan.")
    else:
        display(undocumented)

print("\nNilai repayment status yang ditemukan pada train:")

for column in config.REPAYMENT_STATUS_COLUMNS:
    print(column, sorted(X_train[column].dropna().unique().tolist()))
```

```python
from src import config


def audit_zero_amounts(
    X: pd.DataFrame,
    y: pd.Series,
) -> pd.DataFrame:
    """Summarize rows with all twelve financial amounts equal to zero."""
    if not X.index.equals(y.index):
        raise ValueError("Index fitur dan target harus sama.")

    amount_columns = [
        *config.BILL_COLUMNS,
        *config.PAYMENT_COLUMNS,
    ]
    amounts = X.loc[:, amount_columns]

    all_zero = (
        amounts.notna().all(axis=1)
        & amounts.eq(0).all(axis=1)
    )

    audit = pd.DataFrame(
        {
            "all_amounts_zero": all_zero,
            "target": y,
        },
        index=X.index,
    )

    summary = (
        audit.groupby("all_amounts_zero")["target"]
        .agg(
            jumlah="size",
            jumlah_default="sum",
            default_rate="mean",
        )
    )

    summary["default_rate_pct"] = (
        summary.pop("default_rate") * 100
    )
    summary["share_train_pct"] = (
        summary["jumlah"] / len(X) * 100
    )

    return summary


display(audit_zero_amounts(X_train, y_train).round(2))

amount_columns = [
    *config.BILL_COLUMNS,
    *config.PAYMENT_COLUMNS,
]

zero_amount_mask = (
    X_train[amount_columns].notna().all(axis=1)
    & X_train[amount_columns].eq(0).all(axis=1)
)

display(
    X_train.loc[
        zero_amount_mask,
        [
            "LIMIT_BAL",
            *config.REPAYMENT_STATUS_COLUMNS,
            *amount_columns,
        ],
    ].head(10)
)
```
